Apply transparent, explainable cleaning

Remove quantity anomalies

```python
rows_before = transaction_data.shape[0]

df_clean = transaction_data[transaction_data['PROD_QTY'] <= 10]

print(f"Removed {rows_before - df_clean.shape[0]} rows due to extreme PROD_QTY")
```

Cap extreme sales

```python
sales_cap = df_clean['TOT_SALES'].quantile(0.995)

df_clean['TOT_SALES'] = df_clean['TOT_SALES'].clip(upper=sales_cap)
```

```python
df_clean[['PROD_QTY', 'TOT_SALES']].describe()
```

```python
df_clean[['PROD_QTY', 'TOT_SALES']].quantile([0.95, 0.99])
```

After removing extreme bulk-purchase transactions and validating the distributions, the data shows stable and realistic consumer purchasing behavior, with 99% of transactions involving two units or fewer and sales values aligned with expected product pricing.

### Checking data formats and correcting

Transaction data 

```python
df_clean.info()
```

Customer data

```python
df_customers = pd.read_csv('QVI_purchase_behaviour.csv')

df_customers
```

```python
df_customers.info()
```

Correct formats — Transaction data

DATE → datetime

```python
df_clean['DATE'] = pd.to_datetime(df_clean['DATE'], origin='1899-12-30', unit='D')
```

```python
df_clean['DATE'].min(), df_clean['DATE'].max()
```

```python
# PROD_QTY → integer

df_clean['PROD_QTY'] = df_clean['PROD_QTY'].astype(int)
df_clean['PROD_QTY'].dtype
```

```python
# TOT_SALES → float
df_clean['TOT_SALES'] = df_clean['TOT_SALES'].astype(float)
```

Correct formats — Customer data

```python
# IDs
df_customers['LYLTY_CARD_NBR'] = df_customers['LYLTY_CARD_NBR'].astype(int)
```

Convert categorical columns to category dtype

This improves memory and prevents typos later.

```python
cat_cols = ['LIFESTAGE', 'PREMIUM_CUSTOMER']
df_customers[cat_cols] = df_customers[cat_cols].astype('category')
```

```python
df_customers.dtypes
```

Standardize categorical text

Remove hidden whitespace

```python
df_customers['LIFESTAGE'] = df_customers['LIFESTAGE'].str.strip()
df_customers['PREMIUM_CUSTOMER'] = df_customers['PREMIUM_CUSTOMER'].str.strip()
```

```python
df_customers['LIFESTAGE'].value_counts()
```

```python
df_customers['PREMIUM_CUSTOMER'].value_counts()
```

Validate join key compatibility

This is critical before merging.

```python
df_clean['LYLTY_CARD_NBR'].dtype
df_customers['LYLTY_CARD_NBR'].dtype
```

```python
df_clean.dtypes
```

### derive extra features

extract pack size

```python
df_clean['PACK_SIZE'] = (
    df_clean['PROD_NAME']
    .str.extract(r'(\d+)\s*[gG]', expand=False)
    .astype('Int64')
)
```

```python
# validate
df_clean['PACK_SIZE']
```

extract brand

```python
df_clean['BRAND'] = df_clean['PROD_NAME'].str.split().str[0]
```

Clean brand names

```python
df_clean['BRAND'] = df_clean['BRAND'].replace({
    'Dorito': 'Doritos',
    'Red': 'Red Rock',
    'Infzns': 'Infuzions',
    'Smith': 'Smiths'
})
```

```python
# validate
df_clean['BRAND'].value_counts()
```

Create price-based features

These help explain what drives spend.

```python
df_clean['PRICE_PER_GRAM'] = df_clean['TOT_SALES'] / (df_clean['PROD_QTY'] * df_clean['PACK_SIZE'])
```

Merge customer data

```python
df_merged = df_clean.merge(
    df_customers,
    on='LYLTY_CARD_NBR',
    how='left'
)
```

```python
# validate
df_merged.isna().sum()
```

Define metrics of interest

who spends on chips, and why?

```python
customer_metrics = (
    df_merged
    .groupby(['LYLTY_CARD_NBR', 'LIFESTAGE', 'PREMIUM_CUSTOMER'])
    .agg(
        total_spend=('TOT_SALES', 'sum'),
        avg_spend=('TOT_SALES', 'mean'),
        total_units=('PROD_QTY', 'sum'),
        avg_units=('PROD_QTY', 'mean'),
        avg_pack_size=('PACK_SIZE', 'mean'),
        avg_price_per_gram=('PRICE_PER_GRAM', 'mean'),
        unique_brands=('BRAND', 'nunique'),
        transactions=('TXN_ID', 'nunique')
    )
    .reset_index()
)
```

```python
customer_metrics
```

Segement-level metrics

```python
segment_metrics = (
    customer_metrics
    .groupby(['LIFESTAGE', 'PREMIUM_CUSTOMER'])
    .agg(
        customers=('LYLTY_CARD_NBR', 'nunique'),
        avg_total_spend=('total_spend', 'mean'),
        avg_transactions=('transactions', 'mean'),
        avg_units=('total_units', 'mean'),
        avg_pack_size=('avg_pack_size', 'mean'),
        avg_price_per_gram=('avg_price_per_gram', 'mean')
    )
    .reset_index()
)

segment_metrics
```

Product preference metrics

what drives spend

```python
brand_pref = (
    df_merged
    .groupby(['LIFESTAGE', 'PREMIUM_CUSTOMER', 'BRAND'])
    .agg(
        total_spend=('TOT_SALES', 'sum'),
        units=('PROD_QTY', 'sum')
    )
    .reset_index()
)

brand_pref
```

Premium older families spend more on chips primarily due to purchasing larger pack sizes rather than paying higher prices per gram, indicating volume-driven rather than price-driven spend.

### Visualization

Spend by customer segment (WHO spends more)

```python
# prepare data
segment_spend = (
    df_merged
    .groupby(['LIFESTAGE', 'PREMIUM_CUSTOMER'])
    .agg(
        avg_spend=('TOT_SALES', 'mean'),
        total_spend=('TOT_SALES', 'sum')
    )
    .reset_index()
)
```

```python
# visualization Average spend per transaction
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.barplot(
    data=segment_spend,
    x='LIFESTAGE',
    y='avg_spend',
    hue='PREMIUM_CUSTOMER'
)

plt.title('Average Spend per Transaction by Customer Segment')
plt.ylabel('Average Spend ($)')
plt.xlabel('')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Pack size preference by segment (VOLUME driver)

Understand whether higher spend comes from larger packs, not higher prices.

```python
# prepare data
pack_pref = (
    df_merged
    .groupby(['LIFESTAGE', 'PREMIUM_CUSTOMER', 'PACK_SIZE'])
    .agg(total_spend=('TOT_SALES', 'sum'))
    .reset_index()
)
```

```python
# Visualization: Pack size vs spend
g = sns.catplot(
    data=pack_pref,
    x='PACK_SIZE',
    y='total_spend',
    col='PREMIUM_CUSTOMER',
    row='LIFESTAGE',
    kind='bar',
    height=3,
    aspect=1.2
)

g.set_titles('{row_name} | {col_name}')
g.set_axis_labels('Pack Size (g)', 'Total Spend ($)')
plt.tight_layout()
plt.show()
```

Brand-driven spend (WHAT brands drive spend)

Identify brand preferences by segment.

```python
# prepare data
brand_spend = (
    df_merged
    .groupby(['LIFESTAGE', 'BRAND'])
    .agg(total_spend=('TOT_SALES', 'sum'))
    .reset_index()
)
```

```python
top_brands = (
    brand_spend
    .sort_values('total_spend', ascending=False)
    .groupby('LIFESTAGE')
    .head(5)
)

plt.figure(figsize=(12, 6))
sns.barplot(
    data=top_brands,
    x='BRAND',
    y='total_spend',
    hue='LIFESTAGE'
)

plt.title('Top Brands by Spend per Lifestage')
plt.ylabel('Total Spend ($)')
plt.xlabel('')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Insights

Older families spend more primarily due to larger pack sizes, not higher prices per gram.

Premium customers pay consistently more per gram across all lifestages, indicating lower price sensitivity.

Young singles favor smaller packs but premium brands, suggesting impulse and brand-driven spend.
